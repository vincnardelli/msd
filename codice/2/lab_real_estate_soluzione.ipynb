Un'agenzia immobiliare vuole capire quanto il livello socioeconomico di un quartiere pesa sul valore delle case a Boston.
`boston.csv` descrive 506 quartieri: `medv` (valore mediano delle case, migliaia di \$), `lstat` (% di popolazione a basso reddito), `rm` (stanze medie), `age` (% di case costruite prima del 1940), `dis` (distanza dai centri di lavoro).

```python
import polars as pl
from plotnine import *
import statsmodels.formula.api as smf
```

```python
boston = pl.read_csv("https://www.vincnardelli.com/msd/dati/boston.csv")
boston.head()
```

### 1. Come sono distribuiti `lstat` e `medv`? Istogrammi e misure di sintesi.

```python
boston.select("lstat", "medv").describe()
```

```python
ggplot(boston, aes(x="lstat")) + geom_histogram(bins=30)
```

```python
ggplot(boston, aes(x="medv")) + geom_histogram(bins=30)
```

**Risposta.** `lstat` è asimmetrica a destra: molti quartieri sotto il 15%, pochi sopra il 30%. `medv` ha un picco anomalo a 50: i valori sono stati troncati a 50.000 \$.

### 2. Che relazione c'è tra `lstat` e `medv`? Grafico a dispersione e correlazione.

```python
ggplot(boston, aes(x="lstat", y="medv")) + geom_point()
```

```python
boston.select("lstat", "medv").corr()
```

**Risposta.** Negativa e forte (correlazione −0,74): più povertà nel quartiere, case che valgono meno.

### 3. Stima `medv ~ lstat`. Come leggi intercetta e pendenza? Quanto vale l'$R^2$?

```python
modello = smf.ols("medv ~ lstat", data=boston).fit()
modello.summary()
```

**Risposta.** Intercetta 34,55: valore atteso (34.550 \$) in un quartiere senza popolazione a basso reddito. Pendenza −0,95: ogni punto percentuale in più di `lstat` si associa a 950 \$ in meno di valore. $R^2$ = 0,544: `lstat` spiega il 54% della variabilità.

### 4. Che valore prevedi per quartieri con il 5%, il 10% e il 15% di popolazione a basso reddito?

```python
modello.predict({"lstat": [5, 10, 15]})
```

**Risposta.** Circa 29.800, 25.050 e 20.300 \$.

### 5. Guarda il grafico con la retta: la relazione è davvero una retta?

```python
ggplot(boston, aes(x="lstat", y="medv")) + geom_point() + geom_smooth(method="lm")
```

**Risposta.** No: i punti disegnano una curva. La retta sovrastima il valore nei quartieri a `lstat` intermedio e lo sottostima agli estremi. Una retta è un buon primo passo, non l'ultimo.

### 6. Per chi finisce prima: aggiungi `rm`, `age` e `dis`. Come cambiano coefficienti e $R^2$?

```python
multiplo = smf.ols("medv ~ lstat + rm + age + dis", data=boston).fit()
multiplo.summary()
```

**Risposta.** L'$R^2$ sale a 0,649. Il coefficiente di `lstat` passa da −0,95 a −0,68: una parte del suo effetto era dovuta al numero di stanze, che va insieme alla povertà del quartiere. `age` non è significativa (p = 0,08) a parità delle altre.
