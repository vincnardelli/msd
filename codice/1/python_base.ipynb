## 2. Liste

Una lista è una sequenza di valori tra parentesi quadre. **Attenzione**: in Python si conta da 0.

```python
redditi = [1200, 1500, 980, 2100, 1750]
print(redditi[0])        # primo elemento
print(len(redditi))      # quanti elementi
print(sum(redditi) / len(redditi))   # media
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.formula.api as smf

DATI = "https://www.vincnardelli.com/msd/dati/"
```

## 4. Leggere un dataset

Tutti i dati del corso sono file CSV sul sito: si leggono con `pd.read_csv`.

```python
happy = pd.read_csv(DATI + "happiness.csv")
happy.head()
```

```python
happy.shape        # righe, colonne
```

```python
happy.describe()   # statistiche di base di tutte le colonne numeriche
```

## 5. Selezionare colonne e righe

```python
happy["happyScore"].mean()      # una colonna, poi la sua media
```

```python
happy[happy["happyScore"] > 7]  # solo le righe che rispettano la condizione
```

```python
happy.sort_values("median_income", ascending=False).head(5)   # i 5 paesi con reddito mediano più alto
```

## 6. Un grafico

```python
plt.hist(happy["happyScore"], bins=20)
plt.xlabel("Punteggio di felicità")
plt.ylabel("Frequenza")
plt.show()
```
