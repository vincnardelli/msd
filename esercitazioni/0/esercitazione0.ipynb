Per utilizzare le funzioni di Python, installo i pacchetti di cui ho bisogno. Polars per tabelle e statistiche descrittive

```python
import polars as pl #importo pacchetto per visualizzare tabella
```

```python
happy = pl.read_csv("https://www.vincnardelli.com/msd/dati/happiness.csv") #carico dati dal sito
```

Per utilizzare le funzioni di Python, installo i pacchetti di cui ho bisogno. Plotnine per grafici e visualizzazione

```python
from plotnine import *
```

```python
ggplot(happy, aes(x="median_income"))+geom_histogram()+labs(x="Reddito mediano", title="Istogramma del reddito mediano" )
#i grafici si compongono con +
```

```python
ggplot(happy, aes(x="median_income",y="happyScore"))+geom_point()+labs(x="Reddito mediano", y="indice di felicità")
```
